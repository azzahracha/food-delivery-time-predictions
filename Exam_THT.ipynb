# Data Preprocessing

## Cleaning

```python
# Duplicate Value
df.duplicated().sum()
```

```python
# Missing Value
missing_summary = pd.DataFrame({
    'Type' : df.dtypes,
    'Missing Count': df.isnull().sum(),
    'Missing Percentage (%)': df.isnull().mean() * 100
})

missing_summary = missing_summary[
    missing_summary['Missing Count'] > 0
].sort_values(
    'Missing Percentage (%)',
    ascending=False
)

missing_summary
```

```python
## Visualisasi Distribusi Nilai pada Kolom Courier Experience
import plotly.graph_objects as go

mean_value = df['Courier_Experience_yrs'].mean()
median_value = df['Courier_Experience_yrs'].median()

fig = go.Figure()

fig.add_trace(
    go.Histogram(
        x=df['Courier_Experience_yrs'].dropna(),
        nbinsx=10,
        name='Courier Experience'
    )
)

# Mean
fig.add_vline(
    x=mean_value,
    line_dash='dash',
    annotation_text=f'Mean = {mean_value:.2f}',
    annotation_position='top'
)

# Median
fig.add_vline(
    x=median_value,
    line_dash='dot',
    annotation_text=f'Median = {median_value:.2f}',
    annotation_position='bottom'
)

fig.update_layout(
    title='Distribution of Courier Experience',
    xaxis_title='Courier Experience (years)',
    yaxis_title='Number of Couriers',
    template='plotly_white'
)

fig.show()
```

```python
df['Courier_Experience_yrs'].skew()
```

```python
## Solve - Missing Value pada Kolom Numerik
data = df.copy()
data['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].median())
data.isna().sum()
```

```python
## Solve - Missing Value pada Kolom Kategorik
categorical = ['Weather', 'Traffic_Level', 'Time_of_Day']

for col in categorical:
    data[col] = data[col].fillna('Unknown')

data.isna().sum()
```

```python
# Outlier
numeric_columns = [
    'Distance_km',
    'Preparation_Time_min',
    'Courier_Experience_yrs',
    'Delivery_Time_min'
]

df[numeric_columns].describe().T
```

```python
for col in numeric_columns:

    fig = go.Figure()

    fig.add_trace(
        go.Box(
            y=data[col],
            name=col
        )
    )

    fig.update_layout(
        title=f'Distribution and Outliers - {col}',
        yaxis_title=col,
        template='plotly_white'
    )

    fig.show()
```

```python
# Checking type of data
data.info()
```

## Feature Engineering

```python
# Grouping Berdasarkan Jarak Pengantaran
data['Distance_Group'] = pd.cut(
    data['Distance_km'],
    bins=[-float('inf'), 3, 7, float('inf')],
    labels=['Short', 'Medium', 'Long']
)
print(data['Distance_Group'].value_counts().sort_index())
```

```python
# Grouping Berdasarkan Lama Persiapan Pesanan
data['Preparation_Group'] = pd.qcut(
    data['Preparation_Time_min'],
    q=3,
    labels=['Fast', 'Moderate', 'Slow']
)
print(data['Preparation_Group'].value_counts().sort_index())
```

```python
# Grouping Berdasarkan Pengalaman Kurir
data['Experience_Group'] = pd.cut(
    data['Courier_Experience_yrs'],
    bins=[-1, 2, 5, float('inf')],
    labels=['Beginner', 'Intermediate', 'Experienced']
)
print(data['Experience_Group'].value_counts().sort_index())
```

### Line Chart - Time of Day

```python
time_order = ['Morning', 'Afternoon', 'Evening', 'Night', 'Unknown']

time_count = (
    data['Time_of_Day']
    .value_counts()
    .reindex(time_order, fill_value=0)
)

# Membuat line chart
fig = go.Figure(
    go.Scatter(
        x=time_count.index,
        y=time_count.values,
        mode='lines+markers+text',
        text=time_count.values,
        textposition='top center',
        line=dict(width=3),
        marker=dict(size=8),
        hovertemplate='<b>%{x}</b><br>Number of Orders: %{y}<extra></extra>'
    )
)

# Layout
fig.update_layout(
    title='Number of Orders by Time of Day',
    xaxis_title='Time of Day',
    yaxis_title='Number of Orders',
    height=450,
    width=800,
    margin=dict(l=70, r=70, t=80, b=60)
)

fig.show()
```

```python
Q1 = data['Delivery_Time_min'].quantile(0.25)
Q3 = data['Delivery_Time_min'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Tandai outlier
data['Delivery_Outlier'] = (
    (data['Delivery_Time_min'] < lower_bound) |
    (data['Delivery_Time_min'] > upper_bound)
)

print("Q1 :", Q1)
print("Q3 :", Q3)
print("IQR :", IQR)
print("Lower Bound :", lower_bound)
print("Upper Bound :", upper_bound)

print("\nJumlah outlier :", data['Delivery_Outlier'].sum())
print("Persentase outlier :",
      round(data['Delivery_Outlier'].mean() * 100, 2), "%")
```

```python
normal = data[data['Delivery_Outlier'] == False]
outlier = data[data['Delivery_Outlier'] == True]

fig = go.Figure()

experience_groups = ['Beginner', 'Intermediate', 'Experienced']

for group in experience_groups:

    subset = data[data['Experience_Group'] == group]

    fig.add_trace(
        go.Scatter(
            x=subset['Distance_km'],
            y=subset['Delivery_Time_min'],
            mode='markers',
            name=group,
            text=subset['Order_ID'],
            hovertemplate=
                'Order ID: %{text}<br>' +
                'Experience: ' + group + '<br>' +
                'Distance: %{x:.2f} km<br>' +
                'Delivery Time: %{y:.2f} min' +
                '<extra></extra>'
        )
    )

# Garis batas atas outlier
fig.add_hline(
    y=upper_bound,
    line_dash='dash',
    annotation_text=f'Upper Bound = {upper_bound:.2f} min',
    annotation_position='top right'
)

fig.update_layout(
    title='Distance vs Delivery Time by Courier Experience',
    xaxis_title='Distance (km)',
    yaxis_title='Delivery Time (minutes)',
    template='plotly_white',
    height=600
)

fig.show()
```

### Numerical Analysis
> Correlation Matrix of Numerical Columns

```python
# Heatmap

corr_cols = [
    'Distance_km',
    'Preparation_Time_min',
    'Courier_Experience_yrs',
    'Delivery_Time_min'
]

correlation = data[corr_cols].corr()

fig = go.Figure(
    go.Heatmap(
        z=correlation.values,
        x=correlation.columns,
        y=correlation.columns,
        text=correlation.round(2).values,
        texttemplate='%{text}',
        colorscale='RdBu',
        zmin=-1,
        zmax=1
    )
)

fig.update_layout(
    title='Correlation Matrix of Numerical Variables',
    template='plotly_white'
)

fig.show()
```
